We can "iterate" (AKA loop) over an enumerated list. An example can be found in the following code. **Alter the code to only `print()` one of the following statements for each organism.**

> `organism` is an animal.

> `organism` is not an animal.

Example of desired output: `"squirrel is an animal"`

Hints:

* **Watch the [indentation](https://www.w3schools.com/python/gloss_python_indentation.asp)!** 

* Here's more information on [if statements](https://www.w3schools.com/python/python_conditions.asp)

* To [access a specific item from a list](https://www.w3schools.com/python/python_lists_access.asp), we can use what is called an *index*. **In Python, all indexing starts at 0!** So the first item in a list would be accessed as follows:

    `my_list[0]`

```python
# Enumerating over the list of organisms and printing information

for i, organism in enumerate(organisms):
    
    # Is organism_i an animal?
    
    if is_animal[i]: # Yes!
        
        print('Organism ', i) # print organism index
        print(organism) # print organism
        print(is_animal[i]) # print if organism is an animal
        print('~~~') # Line break
        
    else: # No
        
        print('Organism ', i) # print organism index
        print(organism) # print organism
        print(is_animal[i]) # print if organism is an animal
        print('~~~') # Line break
```
